```python
datos = np.random.randint(1,200, size=(18,))
print("Los datos son:")
print(datos)

minimo = datos.min()
maximo = datos.max()

print(f"\nLos datos tienen un rango de {maximo - minimo}, abarcando desde el {minimo} hasta el {maximo}")
```

```python
minimo = np.amin(datos)
maximo = np.amax(datos)
print(f"\nLos datos tienen un rango de {maximo - minimo}, abarcando desde el {minimo} hasta el {maximo}")
primer_cuartil = np.percentile(datos, 25)
mediana = np.median(datos)
tercer_cuartil = np.percentile(datos, 75)
print(f"\nLa mediana de los datos es igual a {mediana}")
print(f"El primer cuartil es igual a {primer_cuartil}, mientras que el tercer cuartil es igual a {tercer_cuartil}")
print(f"Con rango intercuartil = {primer_cuartil} a {tercer_cuartil} o {tercer_cuartil - primer_cuartil}")
```

```python
media = datos.mean()
desviacion_estandar = datos.std()
varianza = datos.var()
print(f"\nLos datos poseen una media aritmetica de {media:.4f}, con una varianza de {varianza:.4f} y una desviacion estandar de {desviacion_estandar:.4f}")
```

```python
import numpy as np
notas = np.array([78, 92, 65, 88, 70])
promedio = np.mean(notas)
maxima = np.max(notas)
minima = np.min(notas)
cantidad = np.sum(notas >= 70)

print("Promedio:", promedio)
print("Maxima:", maxima)
print("Minima:", minima)
print("Notas ≥ 70:", cantidad)

if promedio >= 70:
  print("El promedio alcanza 70.")
else:
  print("El promedio no alcanza 70.")
```

```python
import numpy as np
temperaturas = np.array([18, 21, 25, 19, 23, 27, 20])
promedio = np.mean(temperaturas)
dias_encima = np.sum(temperaturas > promedio)
fahrenheit = temperaturas * 9/5 + 32

print("Promedio:", promedio)
print("Dias por encima:", dias_encima)
print("Fahrenheit:", fahrenheit)
```
