```python
cola = list() #Cola utilizando una lista
for cantidad in range(5):
  nuevo = input ("Nombre del recien llegado:")
  cola.append(nuevo)

print(f"Se agregaron {len(cola)} elementos:")
for elemento in cola:
  print(elemento)

#Hacer notar que los elementos permacen en la cola
print(f"Actualmente hay {len(cola)} encolados")

print("Procedemos a retirarlos de la cola")

while cola:
  print(cola.pop(0))

#Verificar que la estructura se encuentra vacia
print(f"Actualmente hay {len(cola)} encolados")
```

```python
import collections

cola = collections.deque()

for cantidad in range(5):
  nuevo = input("Nombre del recien llegado: ")
  cola.append(nuevo)

print(f"Se agregaron {len(cola)} elementos:")
for elemento in cola:
  print(elemento)


print("Procedemos a retirarlos de la cola")

while cola:
  print(cola.popleft())

  print(f"Actualmente se encuentra en la cola {len(cola)} elementos")
```
