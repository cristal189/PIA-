```python
import pandas as pd
df = pd.read_csv(
    'https://raw.githubusercontent.com/asalber/manual-python/master/datos/colesterol.csv')

# Creamos una serie con los datos del nuevo registro
nueva_fila = pd.Series(['Carlos Rivas', 28, 'H', 89.0, 1.78, 245.0],
                       index=['nombre', 'edad', 'sexo', 'peso', 'altura', 'colesterol'])

# Usamos pd.concat convirtiendo la Serie en un DataFrame de una fila (transponiéndola)
df = pd.concat([df, nueva_fila.to_frame().T], ignore_index=True)

print(df.tail())
```

```python
import pandas as pd
datos = {'nombre': ['Maria', 'Luis', 'Carmen'],
         'edad': [18, 22, 20],
         'Matematicas': [8.5, 7, 3.5],
         'Economía': [8, 6.5, 5],
         'Programación': [6.5, 4, 9]}
df = pd.DataFrame(datos)
df1 = df.melt(id_vars=['nombre', 'edad'],
              var_name='asignatura', value_name='nota')
print(df1)
```

```python
print(df1.pivot(index='nombre',
                columns='asignatura', values='nota'))
```
