```python
Entradas=[
    ['correo', 'nombre','telefono'],
    ['juan@gmail.com','Juan','8123232323'],
    ['maria@gmail.com','Maria','5545454545'],
    ['diana@homail.com','Diana','4490909090'],
]
for e in Entradas:
  print(f'{e[0]}|{e[1]}|{e[2]}')
```

```python
print(">> Contenido de la lista.\n")
print(Entradas)

# Handle file backup/removal before writing
if os.path.exists("agenda.csv"):
  if os.path.exists("agenda.bak"):
    os.remove("agenda.bak")  # Remove old backup
  os.rename("agenda.csv", "agenda.bak") # Make current csv a backup

with open("agenda.csv", "w+") as f:
  for e in Entradas:
    f.write(f"{e[0]}|{e[1]}|{e[2]}\n")

  f.seek(0) # Move to the beginning of the file to read its content
  print("\n>> Contenido del archivo.\n")
  print(f.read()) # Read the entire content of the file
```

```python
Datos=[]
with open("agenda.csv", "r") as f:
  for linea in f:
    lista=linea.split("|")
    lista[2]=lista[2].replace("\n","")
    Datos.append(lista)

print(">>Contenido de la nueva lista.\n")
print(Datos)
```
